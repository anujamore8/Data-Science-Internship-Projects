```python
# Dataset dimensions
print("Rows and Columns:", df.shape)

# Column names
print("\nColumn Names:")
print(df.columns.tolist())

# Data types and non-null counts
print("\nDataset Information:")
df.info()

# Statistical summary
print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Count each content type
type_counts = df['type'].value_counts()

# Calculate percentages
type_percentage = df['type'].value_counts(normalize=True) * 100

# Display results
print("Content Type Counts:")
print(type_counts)

print("\nContent Type Percentage:")
print(type_percentage.round(2))

# Create bar chart
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='type')

plt.title('Distribution of Movies and TV Shows on Netflix')
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    type_counts,
    labels=type_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Percentage of Movies vs TV Shows on Netflix')
plt.show()
```

```python
# Split multiple countries and count each country separately

country_df = df[df['country'] != 'Unknown'].copy()

country_df['country'] = country_df['country'].str.split(',')

country_exploded = country_df.explode('country')

country_exploded['country'] = country_exploded['country'].str.strip()

# Find top 10 countries
top_countries = country_exploded['country'].value_counts().head(10)

print("Top 10 Countries Producing Netflix Content:")
print(top_countries)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_countries.values,
    y=top_countries.index
)

plt.title('Top 10 Countries Producing Netflix Content')
plt.xlabel('Number of Titles')
plt.ylabel('Country')

plt.tight_layout()
plt.show()
```

```python
# Remove unknown or empty genre values
genre_df = df[df['listed_in'] != 'Unknown'].copy()

# Split multiple genres
genre_df['listed_in'] = genre_df['listed_in'].str.split(',')

# Expand each genre into a separate row
genre_exploded = genre_df.explode('listed_in')

# Remove extra spaces
genre_exploded['listed_in'] = genre_exploded['listed_in'].str.strip()

# Count top 10 genres
top_genres = genre_exploded['listed_in'].value_counts().head(10)

print("Top 10 Netflix Genres:")
print(top_genres)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_genres.values,
    y=top_genres.index
)

plt.title('Top 10 Most Popular Genres on Netflix')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')

plt.tight_layout()
plt.show()
```

```python
# Convert date_added into datetime
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

# Extract year
df['year_added'] = df['date_added'].dt.year

print("Date conversion completed!")
print(df[['title', 'date_added', 'year_added']].head())
```

```python
content_by_year = df['year_added'].value_counts().sort_index()

print("Content Added by Year:")
print(content_by_year)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(content_by_year.index, content_by_year.values, marker='o')

plt.title('Netflix Content Added Over the Years')
plt.xlabel('Year')
plt.ylabel('Number of Titles Added')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
yearly_type = df.groupby(['year_added', 'type']).size().unstack(fill_value=0)

yearly_type.plot(figsize=(12, 6), marker='o')

plt.title('Movies vs TV Shows Added Over the Years')
plt.xlabel('Year')
plt.ylabel('Number of Titles')
plt.legend(title='Content Type')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Count content ratings
rating_counts = df['rating'].value_counts()

print("Content Rating Distribution:")
print(rating_counts)
```

```python
top_ratings = rating_counts.head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_ratings.values,
    y=top_ratings.index
)

plt.title('Top 10 Most Common Netflix Content Ratings')
plt.xlabel('Number of Titles')
plt.ylabel('Content Rating')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 6))

sns.countplot(
    data=df,
    x='rating',
    hue='type',
    order=df['rating'].value_counts().index
)

plt.title('Content Ratings: Movies vs TV Shows')
plt.xlabel('Content Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
